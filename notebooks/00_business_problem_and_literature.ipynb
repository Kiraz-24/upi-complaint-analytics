# 00 · Business problem and literature support

**Guidelines covered:** Business problem definition · Literature support

## Business problem

### Context
The Unified Payments Interface (UPI) is India's real-time payment system: any bank account can pay any other in seconds from a phone, by scanning a QR code or entering a UPI ID. It processed about **24.5 billion transactions in August 2026 alone** (National Payments Corporation of India). Almost all of this volume flows through a handful of apps: PhonePe, Google Pay, Paytm and the government-backed BHIM.

When these apps fail (money is deducted but not received, the app will not open, the login code never arrives, a security check locks out a genuine user), customers complain in public, in Google Play Store reviews. Each app receives thousands of reviews a day. No product team can read them all, so priorities are often set by gut feeling, by the loudest complaint, or by internal metrics that do not show how customers experience the problem.

### Decision maker and business question
- **Decision maker:** the Head of Product or Head of Customer Experience of a UPI app, who decides what goes into the next development sprint.
- **Business question:** *Which failures drive 1-star reviews for BHIM, Google Pay, Paytm and PhonePe, and which fix would lift each app's rating most?*

### Why it matters
- **Ratings drive adoption.** App-store ratings are strongly linked to downloads (Harman, Jia and Zhang, 2012), and negative reviews weigh more than positive ones (Chevalier and Mayzlin, 2006).
- **Switching costs are near zero.** All UPI apps connect to the same bank accounts, so an unhappy user can move to a competitor in minutes.
- **Complaints are public.** Every unresolved complaint is visible to every potential new user.

### Objectives
| # | Objective | Notebook |
|---|---|---|
| O1 | Collect a large, current, self-scraped dataset of reviews for the four apps | 01 |
| O2 | Identify complaints reliably, including complaints hidden behind high star ratings | 05 |
| O3 | Discover what users complain about, per app, without predefined categories | 06 |
| O4 | Quantify how many stars each problem costs and rank fixes by expected rating gain | 07 |
| O5 | Detect problematic app releases and emerging problems early | 07, 08 |
| O6 | Deliver the results as a decision tool a product team can use every week | 08 |

### Scope
Four apps; 86,056 English-interface Google Play reviews from May to October 2026; Android only (more than 90% of Indian smartphones). The project measures *association* between complaint themes and ratings; it does not have access to internal company data such as crash logs or support tickets.

### Success criteria
1. A sentiment model that clearly beats a lexicon baseline (macro F1).
2. Complaint themes that are coherent (topic coherence) and interpretable by a manager.
3. Statistically tested, quantified recommendations per app.
4. An interactive dashboard that answers the business question without code.

## Literature support

### 1. App-store reviews as customer intelligence
App-store reviews are one of the largest sources of unsolicited customer feedback. Pagano and Maalej (2013) showed that a large share of app-store feedback contains information that is directly useful for product decisions (bug reports, feature requests, user experience) and that most feedback arrives shortly after new releases. Harman, Jia and Zhang (2012) found a strong correlation between an app's customer rating and its download rank, which links review sentiment to commercial performance. A systematic review of 182 studies by Dąbrowski et al. (2022) confirms that review mining is now an established research field, but notes that most work is aimed at software engineers (classifying reviews into bug reports or feature requests) rather than at business decision makers. Jha and Mahmoud (2019) showed that about 40% of reviews in their sample raised a non-functional requirement such as security, performance or usability, the same kind of quality problems that dominate UPI-app complaints. Alves de Lima et al. (2022) analysed around 85,000 reviews and showed that the frequency of negative opinions about a feature can predict future rating problems, which supports this project's release monitor and emerging-topic analysis.

### 2. Why ratings matter commercially
Chevalier and Mayzlin (2006) showed with online book reviews that better reviews raise relative sales and that very negative reviews have a stronger effect than very positive ones. Hu and Liu (2004) established the idea of summarising large volumes of customer reviews by product feature, which is the conceptual basis for turning thousands of reviews into a short list of problem themes. Together these studies justify the business premise of this project: complaint themes that cost stars are worth fixing because ratings influence adoption.

### 3. The UPI context
UPI is a real-time, interoperable, mobile-first payment system operated by the National Payments Corporation of India (Gochhwal, 2017). Adoption research shows that relative advantage, complexity and observability drive the intention to use UPI (Fahad, 2022), and that trust shapes users' attitude towards UPI (Razi-ur-Rahim et al., 2024). These survey-based studies explain why people adopt UPI, but not what goes wrong after adoption. Review analytics fills this gap with behavioural evidence at scale: the problems real users report, app by app and release by release.

### 4. Sentiment analysis methods
Lexicon-based tools such as VADER (Hutto and Gilbert, 2014) score text with a hand-built dictionary of emotional words and rules for negation and intensity. They need no training data and work well on emotional social-media text, which makes VADER a natural baseline. Transformer models (Vaswani et al., 2017) changed the field: BERT (Devlin et al., 2019) learns word meaning from context in both directions and can be fine-tuned on a specific task, and DistilBERT (Sanh et al., 2019) keeps most of BERT's accuracy while being about 40% smaller and 60% faster. Sentence-BERT (Reimers and Gurevych, 2019) and compact MiniLM encoders (Wang et al., 2020) produce sentence embeddings that capture meaning, used here both as a classifier input and for topic modelling. Reviews of Indian apps often mix Hindi and English; the SemEval-2020 SentiMix task (Patwa et al., 2020) showed that sentiment analysis of code-mixed Hinglish text is a distinct and harder problem, which motivated this project's Hinglish normalisation step. Because the classes are imbalanced, model quality is reported with macro-averaged F1 rather than accuracy (Sokolova and Lapalme, 2009).

### 5. Topic modelling and text exploration
Latent Dirichlet Allocation (LDA; Blei, Ng and Jordan, 2003) represents each document as a mixture of word-based topics and is the classic baseline. BERTopic (Grootendorst, 2022) instead clusters documents by meaning: it embeds them, reduces dimensions with UMAP (McInnes, Healy and Melville, 2018), finds dense clusters with HDBSCAN (Campello, Moulavi and Sander, 2013) and describes each cluster with class-based TF-IDF. Comparative studies on financial-services text report clearer topics from embedding-based models than from LDA (Ogunleye et al., 2023). Topic quality is measured with coherence based on normalised pointwise mutual information (NPMI; Bouma, 2009), which Röder, Both and Hinneburg (2015) found to agree well with human judgements. For exploratory text analysis, the weighted log-odds ratio with an informative Dirichlet prior (Monroe, Colaresi and Quinn, 2008) identifies the words that most distinguish two groups of texts, such as complaints and praise.

### 6. Statistical inference for business conclusions
Business recommendations need evidence that a difference is not due to chance. Rating-driver effects are estimated with ordinary least squares and heteroskedasticity-consistent (HC3) standard errors (MacKinnon and White, 1985), because the spread of star ratings differs between groups. Differences in rating distributions between app versions are tested with the Mann-Whitney U test (Mann and Whitney, 1947), which does not assume normally distributed data, and differences in proportions with chi-square tests. Two classifiers evaluated on the same test set are compared with McNemar's test (McNemar, 1947), the approach recommended by Dietterich (1998) for this situation.

### 7. Research gap and contribution
Existing review-mining research is mostly written for software engineers and stops at classification, while UPI research is mostly survey-based adoption studies. Very few studies (i) analyse Indian payment apps with code-mixed reviews, (ii) translate complaint themes into a measured star-rating cost and a fix priority per app, or (iii) monitor app releases statistically. This project combines a fine-tuned transformer, BERTopic and statistical testing into a decision tool for product managers, with a business rather than a purely technical focus.
